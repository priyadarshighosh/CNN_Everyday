```python
import numpy as np

# defie a grayscale image
image = np.array([
    [ 1,2,3,0,1],
    [0,1,2,3,2] ,
    [3,0,1,2,3],
    [2,1,3,0,0],
    [1,2,0,1,2]

])

image.shape
```

```python


# Define a 3 x 3 kernel (filter)
kernel = np.array([
    [1,0,-1],
    [1,-4,1],
    [1,0,-1]
])

# define the convolutions
def convolve (image , kernel ):         # image is the image that we pass and kernel s the function that e gonna perform on the image

   image_h , image_w  =image.shape
   kernel_h , kernel_w = kernel.shape

   output_h = image_h - kernel_h + 1      # height wise
   output_w = image_w - kernel_w + 1      # breath wise

   output = np.zeros((output_h , output_w))

   for i in range(output_h):
    for j in range(output_w):

      output[i, j] = np.sum(image[i:i+kernel_h , j:j+kernel_w] * kernel)

   return output


```

```python
# applying the convolution
output = convolve(image , kernel)
print ('Convolution Output:\n', output)
                                                          # we have 5x5 image , 3x3 kernel - we get convolutional output at 3x3
```

# Padding

```python
# Adding Padding
def pad_image(image , pad):
  return np.pad(image, pad , mode='constant', constant_values=0)

pad_image(image , 1 )    # we get a paddded image that is 0 on all the boundary
```

```python


# Modify convolution to include to include stride and padding
def convolve(image, kernel, stride=1, padding=0):    # we have stride 1 but we are using the same logic as above

  if padding > 0 :
    image = pad_image(image , padding)

  image_h , image_w = image.shape
  kernel_h , kernel_w = kernel.shape
  output_h = (image_h - kernel_h) // stride + 1
  output_w = (image_w - kernel_w) // stride + 1
  output = np.zeros((output_h , output_w))

  for i in range(0 , output_h* stride , stride):
    for j in range(0 , output_w * stride , stride):
      output[ i//stride , j //stride] = np.sum(

         image[i:i+kernel_h , j:j+kernel_w] * kernel
      )
  return output




```

```python
image = np.array([
    [ 1,2,3,0,1],
    [0,1,2,3,2] ,
    [3,0,1,2,3],
    [2,1,3,0,0],
    [1,2,0,1,2]

])

# The function you defined is called 'convolve', not 'convolve_with_padding_and_stride'
padded_output = convolve(image, kernel, stride=1, padding=1)
print("Convolution Output with Padding:\n", padded_output)
```

# Max Pooling

```python
def max_pool(image , pool_size , stride):
  image_h , image_w = image.shape
  output_h = (image_h - pool_size) // stride + 1     # if we have a 6x6 image with 2x2 pool we get a 3x3 image
  output_w = (image_w - pool_size) // stride + 1     # as output ( we are doing floor operation here )
  output = np.zeros((output_h , output_w))

  for i in range(0 , output_h * stride ,stride):
    for j in range(0 , output_w * stride , stride):
      output[i//stride , j//stride] = np.max(
          image[i:i+pool_size , j:j+pool_size]
      )
  return output
```

```python
image = np.array([
    [1,2,3,0,1],
    [0,1,2,3,2],
    [3,0,1,2,1],
    [2,1,3,0,0],
    [1,2,0,1,2]
                  ])

# apply pool maxing
output = max_pool(image , pool_size=2 , stride=2)
print("Max Pooling Output:\n", output)
```
